## Setup (run once in a terminal at repo root)
```bash
uv sync                       # creates .venv with Python 3.12 + base deps
uv run python -m ipykernel install --user --name ai-eng   # kernel for Jupyter
cp .env.example .env          # then fill in the keys you have
```
Pick the **ai-eng** kernel (or `.venv`) in your editor.

Adding libraries as lessons need them:
```bash
uv add torch scikit-learn     # updates pyproject.toml + uv.lock
```
Inside a notebook you can also run `!uv add <pkg>` and restart the kernel.

## Alternatives to try once
- `python -m venv .venv && source .venv/bin/activate && pip install -r ...` — the classic way you'll see in most tutorials.
- `conda create -n ai python=3.12` — what many data-science courses assume.


## Check your environment

````python
import importlib.util, platform, sys

print("python  ", sys.version.split()[0], "| executable:", sys.executable)
print("platform", platform.platform(), platform.machine())
assert sys.prefix != sys.base_prefix, "Not running inside a virtualenv: select the ai-eng / .venv kernel"

for pkg in ["numpy", "pandas", "matplotlib", "openai", "dotenv"]:
    print(f"{pkg:10}", "ok" if importlib.util.find_spec(pkg) else "MISSING (run `uv sync`)")

if importlib.util.find_spec("torch"):
    import torch
    print("torch", torch.__version__, "| cuda:", torch.cuda.is_available(), "| mps:", torch.backends.mps.is_available())
else:
    print("torch not installed yet (chapter 03 adds it)")
````
