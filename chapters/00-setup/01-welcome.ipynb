## What is an AI engineer?
Someone who **builds products on top of foundation models**. Compared to neighbouring roles:

| Role | Main job | Main tools |
|---|---|---|
| Data scientist | Answer questions with data | pandas, stats, notebooks |
| ML engineer | Train & ship custom models | PyTorch, feature stores, MLOps |
| **AI engineer** | Ship reliable apps on LLMs: prompts, RAG, agents, evals, deployment; fine-tune when needed | LLM APIs, vector DBs, agent frameworks, eval & tracing tools |
| Research engineer | Push model capability | large-scale training, kernels |

AI engineering still needs ML/DL fundamentals (chapters 01-04): you can't debug what you don't understand.

## How this repo works
- One notebook per lesson, under `chapters/NN-chapter/`.
- Each lesson: **Goals** (checkboxes), **Resources** (free first), **Notes** (you write them), **Exercises**.
- Rule: *build > watch*. Every resource is followed by an exercise.
- Milestones (see `README.md`) are the real progress markers. Each one is a project you can show.
- `STACK.md` maps every layer of the stack and its main alternatives. Lessons make you try several of them.

## Study loop
1. Read the goals. 2. Do the resources. 3. Write notes in your own words. 4. Do the exercises. 5. Tick the goals.
